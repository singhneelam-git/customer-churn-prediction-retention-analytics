## Create Customer Power BI Master Table

The customer dataset and machine learning prediction dataset are combined to create a single customer-level reporting table.

This master table will contain both historical customer information and predicted churn risk.

The table will be used as the main data source for Power BI customer analysis.

```python
customer_powerbi = customer_df.merge(
    customer_risk,
    on="customer_id",
    how="left"
)

customer_powerbi.head()
```

```python
customer_powerbi.shape
```

## Validate Customer Risk Information

After merging the customer dataset with the machine learning predictions, we will verify that the churn probability and risk classification fields have been added correctly.

This ensures the Power BI customer table contains both historical and predictive information.

```python
customer_powerbi[
    [
        "customer_id",
        "churn",
        "churn_probability",
        "actual_churn",
        "risk_category"
    ]
].head()
```

## Validate Customer Risk Distribution

The customer master table now contains machine learning risk predictions.

```python
customer_powerbi["risk_category"].value_counts()
```

## Calculate Dashboard KPI Metrics

The Power BI dashboard will include key performance indicators (KPIs) that provide an executive overview of customer churn performance.

These metrics summarise customer retention, financial impact, and customer risk exposure.

```python
total_customers = customer_powerbi["customer_id"].nunique()

churned_customers = customer_powerbi["churn"].sum()

churn_rate = (churned_customers / total_customers) * 100

total_monthly_revenue = customer_powerbi["monthly_revenue"].sum()

high_risk_customers = (
    customer_powerbi["risk_category"] == "High Risk"
).sum()

revenue_at_risk = customer_powerbi.loc[
    customer_powerbi["risk_category"] == "High Risk",
    "monthly_revenue"
].sum()


print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers)
print("Churn Rate:", round(churn_rate, 2), "%")
print("Total Monthly Revenue: £", total_monthly_revenue)
print("High Risk Customers:", high_risk_customers)
print("Revenue At Risk: £", revenue_at_risk)
```

## Export Final Power BI Customer Master Table
